# Examples

Jupyter notebooks in this section are executed when the documentation is built. Download a notebook from its page to run it yourself.

- [Output format](/takeout/examples/output-format/) builds a run in the shared layout, validates it, plots it and stores it as netCDF.
- [Field maps](/takeout/examples/field-maps/) samples a magnetic mirror as a field map, interpolates it inside and outside the grid and reads it back from a file.
- [Scripts](/takeout/example-scripts/) are complete Python scripts with their plots.
- [Building the docs](/takeout/development/docs/) explains how to add an example.
